````python

# Practical 4C: CBOW Model for Textual Document

# 1. Import libraries
import re
from gensim.models import Word2Vec

# 2. Read document
with open("document.txt", "r", encoding="utf-8") as file:
    text = file.read()

# 3. Clean and tokenize text
text = re.sub(r"[^a-z\s]", "", text.lower())
words = text.split()

# Word2Vec needs a list of sentences
sentences = [words]

print("Total Words:", len(words))
print("Sample Words:", words[:20])

if not words:
    raise ValueError("Document is empty!")

# 4. Train CBOW model
model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    sg=0,
    epochs=100
)

print("\nCBOW Model Trained Successfully!")

# 5. Display word embedding
word = words[0]

print("\nSelected Word:", word)
print("Word Embedding:", model.wv[word])

# 6. Find similar words
print("\nSimilar Words:")
print(model.wv.most_similar(word))

# 7. Save model
model.save("cbow_document.model")

print("\nModel Saved Successfully!")
````

For example:

Suppose the sentence is:

`I love machine learning`

If `machine` is the target word, CBOW uses the surrounding words (`I`, `love`, `learning`) to predict `machine`.

## 2. Code explanation step by step

### Step 1: Import libraries

```
import refrom gensim.models import Word2Vec
```

- `re`: Used to remove special characters and numbers.
- `Word2Vec`: Used to train the CBOW model and generate word embeddings.

### Step 2: Read the document

```
with open("document.txt", "r", encoding="utf-8") as file:    text = file.read()
```

Here:

- `open()` opens the text file.
- `"r"` means read mode.
- `encoding="utf-8"` supports different text characters.
- `file.read()` reads the entire document and stores it in `text`.

Example:

If document.txt contains:

`I love machine learning!`

Then `text` stores the complete sentence.

### Step 3: Clean and tokenize text

```
text = re.sub(r"[^a-z\s]", "", text.lower())words = text.split()
```

This step prepares the document for training.

First line:

```
text.lower()
```

Converts all letters into lowercase.

```
re.sub(r"[^a-z\s]", "", text.lower())
```

Removes:

- Numbers
- Special characters
- Punctuation marks

Second line:

```
words = text.split()
```

Splits the document into individual words.

Example:

Before:

`I Love Machine Learning!`

After:

```
['i', 'love', 'machine', 'learning']
```

### Step 4: Prepare sentences

```
sentences = [words]
```

This line is important.

Word2Vec expects data in the form of a list of sentences, where each sentence is itself a list of words.

For example:

```
sentences = [    ['i', 'love', 'machine', 'learning']]
```

Here, we have one document treated as one sentence for training.

```
print("Total Words:", len(words))print("Sample Words:", words[:20])
```

- `len(words)` counts the total words.
- `words[:20]` displays the first 20 words.

```
if not words:    raise ValueError("Document is empty!")
```

Checks whether the document contains any words. If it is empty, the program stops and displays an error.

### Step 5: Train the CBOW model

```
model = Word2Vec(    sentences=sentences,    vector_size=100,    window=5,    min_count=1,    sg=0,    epochs=100)
```

This is the most important part of the practical.

| Parameter         | Meaning                                           |
| ----------------- | ------------------------------------------------- |
| `sentences`       | Input words for training                          |
| `vector_size=100` | Each word is represented using 100 numbers        |
| `window=5`        | Considers surrounding words within a window of 5  |
| `min_count=1`     | Keeps words appearing at least once               |
| `sg=0`            | Uses CBOW architecture                            |
| `epochs=100`      | Trains the model for 100 iterations over the data |

Understand `sg=0`:

- `sg=0` → CBOW
- `sg=1` → Skip-gram

Why 100 epochs? It allows the model to go through the small document multiple times and learn word relationships. However, more epochs do not automatically mean better results.

### Step 6: Display word embedding

```
word = words[0]print("Selected Word:", word)print("Word Embedding:", model.wv[word])
```

- `words[0]` selects the first word.
- `model.wv[word]` retrieves its learned word vector.

For example, the word `machine` may have a vector containing 100 numerical values.

These numbers represent the word in a mathematical form.

### Step 7: Find similar words

```
print(model.wv.most_similar(word))
```

Finds words whose learned vectors are closest to the selected word.

For example, if the selected word is `machine`, it may return related words depending on the document.

Remember: Similarity depends on the training text. It does not guarantee that the words are synonyms.

### Step 8: Save the model

```
model.save("cbow_document.model")
```

Saves the trained CBOW model into a file.

You can load it later without training again:

```
from gensim.models import Word2Vecmodel = Word2Vec.load("cbow_document.model")
```

## 3. Important viva questions

Q1. What is CBOW?

CBOW is a Word2Vec model that predicts a target word using its surrounding context words.

Q2. What is the purpose of this practical?

To learn word relationships from a textual document and convert words into numerical vectors.

Q3. What is Word2Vec?

Word2Vec is a technique that converts words into numerical representations called embeddings.

Q4. What is tokenization?

Breaking text into smaller units, usually individual words.

Q5. Why do we use `sentences = [words]`?

Because Word2Vec expects a list of sentences, with each sentence represented as a list of words.

Q6. What is `vector_size=100`?

It means every word is represented using 100 numerical values.

Q7. What is the difference between CBOW and Skip-gram?

- CBOW: Context words predict the target word.
- Skip-gram: Target word predicts surrounding words.

Q8. What is `model.wv[word]`?

It retrieves the numerical embedding of a particular word.

Q9. What is `most_similar()`?

It finds words with similar vector representations.

Q10. Why do we use epochs?

Epochs determine how many times the model goes through the training data.

Important practical tip

Your code is correct for learning the basic CBOW workflow. One limitation: because the entire document is stored as a single sentence (`sentences = [words]`), CBOW may learn relationships across paragraph boundaries. For a larger document, splitting it into actual sentences before training would be better. Also, a very short document will not produce meaningful embeddings, even if training completes successfully.
