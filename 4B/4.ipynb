````python

# Practical 4B: CBOW Model for Twitter Dataset

# 1. Import libraries
import pandas as pd
import re
from gensim.models import Word2Vec

# 2. Load dataset
data = pd.read_csv("twitter_dataset.csv")

print("Dataset Columns:", data.columns.tolist())

# 3. Select tweet column automatically
possible_columns = ["text", "tweet", "tweets", "content"]

tweet_column = next(
    (col for col in data.columns if col.lower() in possible_columns),
    data.columns[-1]
)

tweets = data[tweet_column].dropna()

print("Selected Column:", tweet_column)

# 4. Clean and tokenize tweets
sentences = []

for tweet in tweets:
    tweet = re.sub(r"[^a-z\s]", "", str(tweet).lower())
    words = tweet.split()

    if words:
        sentences.append(words)

# 5. Check processed data
print("Total Tweets:", len(sentences))

if not sentences:
    raise ValueError("No valid tweets found. Check your dataset column.")

print("Sample Tweet:", sentences[0])

# 6. Train CBOW Word2Vec model
model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    sg=0,
    epochs=10
)

print("\nCBOW Model Trained Successfully!")

# 7. Display word embedding
word = list(model.wv.key_to_index.keys())[0]

print("\nWord:", word)
print("Word Embedding:", model.wv[word])

# 8. Find similar words
print("\nSimilar Words:")
print(model.wv.most_similar(word))

# 9. Save model
model.save("cbow_twitter.model")

print("\nModel Saved Successfully!")
````

## 1. What is CBOW?

CBOW stands for Continuous Bag of Words.

It is a Word2Vec model that predicts a missing or target word using the surrounding words (context).

For example:

`I love machine learning`

If the target word is `machine`, CBOW uses the surrounding words `I`, `love`, `learning` to predict it.

## 3. Viva questions

Q1. What is CBOW?

CBOW is a Word2Vec model that predicts a target word using surrounding context words.

Q2. What is Word2Vec?

Word2Vec is a technique that converts words into numerical vectors called word embeddings.

Q3. What is the difference between CBOW and Skip-gram?

- CBOW predicts the target word from surrounding words.
- Skip-gram predicts surrounding words from the target word.

Q4. What is tokenization?

Breaking a sentence or tweet into individual words.

Q5. What is an embedding?

A numerical representation of a word in vector form.

Q6. What does `sg=0` mean?

It tells Word2Vec to use the CBOW architecture.

Q7. What does `window=5` mean?

It considers surrounding words within a window of five words.

Q8. Why do we clean tweets?

To remove unwanted characters and make the text suitable for training.

Q9. Why do we use `most_similar()`?

To find words that have similar vector representations.

Q10. Why do we save the model?

To reuse the trained model without training it again.
